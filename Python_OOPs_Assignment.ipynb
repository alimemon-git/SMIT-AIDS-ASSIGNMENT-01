**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self, initial_balance=0.0):
        self.__balance = initial_balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit amount positive hona chahiye.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdraw amount positive hona chahiye.")
        elif amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insolvent balance! Aap ke account mein itne paise nahi hain.")


account = BankAccount(1000)

account.deposit(500)
print(account.get_balance())

account.withdraw(300)
print(account.get_balance())

account.withdraw(1500)
print(account.get_balance())
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def __init__(self, species_type):
        self.species_type = species_type

    def eat(self):
        return "Yeh khana kha raha hai."


class Mammal(Animal):
    def __init__(self, species_type, has_fur):
        super().__init__(species_type)
        self.has_fur = has_fur

    def walk(self):
        return "Yeh chal sakta hai."


class Dog(Mammal):
    def __init__(self, species_type, has_fur, breed):
        super().__init__(species_type, has_fur)
        self.breed = breed

    def bark(self):
        return "Woof! Woof!"


my_dog = Dog("Mammal", True, "German Shepherd")

print(my_dog.species_type)
print(my_dog.has_fur)
print(my_dog.breed)

print(my_dog.eat())
print(my_dog.walk())
print(my_dog.bark())
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
#Hierarchical Inheritance mein aik single parent class hoti hai aur us se ek se zyada child classes derive (banayi) ki jati hain. 
#Yani parent aik hi hota hai lekin uske bachay (subclasses) multiple hote hain, aur har child class apne parent ki properties ko inherit karti hai.
class Shape:
    def __init__(self, color):
        self.color = color

    def display_color(self):
        return f"Is shape ka rang {self.color} hai."


class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area_circle(self):
        return 3.14 * self.radius * self.radius


class Square(Shape):
    def __init__(self, color, side):
        super().__init__(color)
        self.side = side

    def area_square(self):
        return self.side * self.side


circle_obj = Circle("Red", 5)
square_obj = Square("Blue", 4)

print(circle_obj.display_color())
print(circle_obj.area_circle())

print(square_obj.display_color())
print(square_obj.area_square())
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
#Operator Overloading ka matlab hai built-in operators (jaise +, -, *, /) ko yeh sikhana ke unho ne kisi custom class ke objects ke sath kaise behave karna hai.
#Normal halat mein, Python do alag objects ko + karne par error deta hai kyunki use nahi pata hota ke unhein kaise jorna hai. Is masle ko hal karne ke liye Python humein Magic Methods (ya Dunder Methods) deta hai. Jab hum apni class mein in special methods ko define karte hain, toh hum un operators ka matlab badal dete hain.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Point(new_x, new_y)

    def display(self):
        return f"({self.x}, {self.y})"


p1 = Point(2, 4)
p2 = Point(3, 5)

p3 = p1 + p2

print(p3.display())
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15
class Calculator:
    def add(self, a=None, b=None, c=None):
        if a is not None and b is not None and c is not None:
            return a + b + c
        elif a is not None and b is not None:
            return a + b
        elif a is not None:
            return a
        else:
            return 0


calc = Calculator()

print(calc.add(5))
print(calc.add(5, 10))
print(calc.add(5, 10, 15))
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17
from abc import ABC, abstractmethod


class Computer(ABC):

    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):

    def process(self):
        return "Laptop process kar raha hai: Portable aur battery par chal raha hai."


class Desktop(Computer):

    def process(self):
        return "Desktop process kar raha hai: Heavy power aur zyada cooling ke sath."


my_laptop = Laptop()
my_desktop = Desktop()

print(my_laptop.process())
print(my_desktop.process())
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod


class Car(ABC):

    def start_engine(self):
        return "Engine start ho gaya hai. Vroom Vroom!"

    @abstractmethod
    def drive(self):
        pass


class Sedan(Car):

    def drive(self):
        return "Sedan sarak par smooth chal rahi hai."


my_car = Sedan()

print(my_car.start_engine())
print(my_car.drive())
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
# Write your code here for Q19
Abstraction aur Encapsulation dono OOP ke bunyadi concepts hain, lekin dono ka maqsad bilkul alag hai:

Abstraction ka maqsad chupana (Hiding complexity) hai. Yeh sirf zaroori details user ko dikhata hai aur andar ki mushkil working ko parde ke peeche rakhta hai. (Yani, kya kaam ho raha hai).

Encapsulation ka maqsad bandhna (Data binding/Restricting access) hai. Yeh data (variables) aur us par kaam karne wale code (methods) ko aik single unit (class) mein lapet deta hai taake bahar se koi data ko ghalat tareeqe se tabdeel na kar sake. (Yani, data ko kaise mehfooz rakhna hai).

Short Definitions
Abstraction: Unnecessary details ko hide karke sirf zaroori aur relevant features ko user ke samne lana. (Example: Gaari chalate waqt break dabana, bina yeh jaane ke andar brake pads kaise kaam kar rahe hain).

Encapsulation: Data aur methods ko aik capsule (class) mein band karna aur private variables ke zariye unauthorized direct access ko rokna. (Example: Bank account ka balance private hona aur usay sirf deposit/withdraw methods ke zariye hi badal sakna).
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod


class Payment(ABC):

    def __init__(self, amount):
        self.amount = amount

    def print_receipt(self):
        print(f"Receipt: Amount Rs.{self.amount} successfully processed.")

    @abstractmethod
    def validate_payment(self):
        pass

    @abstractmethod
    def process_payment(self):
        pass


class CreditCardPayment(Payment):

    def __init__(self, amount, card_number, cvv):
        super().__init__(amount)
        self.card_number = card_number
        self.__cvv = cvv

    def validate_payment(self):
        if len(self.card_number) == 16 and len(str(self.__cvv)) == 3:
            return True
        return False

    def process_payment(self):
        if self.validate_payment():
            print(
                f"Credit Card ({self.card_number[-4:]}) ke zariye Rs.{self.amount} ki payment ho rahi hai..."
            )
            self.print_receipt()
        else:
            print("Credit Card details invalid hain! Payment fail ho gayi.")


class UPIPayment(Payment):

    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def validate_payment(self):
        if "@" in self.upi_id:
            return True
        return False

    def process_payment(self):
        if self.validate_payment():
            print(
                f"UPI ID ({self.upi_id}) ke zariye Rs.{self.amount} ki payment ho rahi hai..."
            )
            self.print_receipt()
        else:
            print("UPI ID invalid hai! Payment fail ho gayi.")


card_pay = CreditCardPayment(5000, "1234567890123456", 123)
card_pay.process_payment()

print("-" * 30)

upi_pay = UPIPayment(1200, "user@bank")
upi_pay.process_payment()
```
