### 43. Ek simple calculator banayein (+, -, *, /) symbols ke basis par operation perform karein.

```python
# Write your code here
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print(f"Result: {num1 + num2}")
elif operator == "-":
    print(f"Result: {num1 - num2}")
elif operator == "*":
    print(f"Result: {num1 * num2}")
elif operator == "/":
    if num2 != 0:
        print(f"Result: {num1 / num2}")
    else:
        print("Error! Division by zero is not allowed.")
else:
    print("Invalid Operator!")
```

### 44. User ka weight aur height le kar BMI category batayein (Underweight, Normal, Overweight).

```python
# Write your code here
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))

bmi = weight / (height ** 2)

if bmi < 18.5:
    print(f"BMI: {bmi:.2f} - Category: Underweight")
elif 18.5 <= bmi < 24.9:
    print(f"BMI: {bmi:.2f} - Category: Normal")
else:
    print(f"BMI: {bmi:.2f} - Category: Overweight")
```

### 54. Ek triangle ki teen sides input lein aur batayein ke wo Equilateral, Isosceles ya Scalene hai.

```python
# Write your code here
side1 = float(input("Enter length of first side: "))
side2 = float(input("Enter length of second side: "))
side3 = float(input("Enter length of third side: "))

if side1 == side2 == side3:
    print("Equilateral Triangle (All sides are equal)")
elif side1 == side2 or side2 == side3 or side1 == side3:
    print("Isosceles Triangle (Two sides are equal)")
else:
    print("Scalene Triangle (No sides are equal)")
```

### 58. Electricity Bill: Pehle 100 units free, next 100 units Rs 5/unit, uske baad Rs 10/unit.

```python
# Write your code here
units = int(input("Enter units consumed: "))

bill = 0

if units <= 100:
    bill = 0
    print("First 100 units are free.")
elif units <= 200:
    bill = (units - 100) * 5
else:
    bill = (100 * 5) + (units - 200) * 10

print(f"Total Electricity Bill: Rs {bill}")
```

### 60. Final Challenge: Ek mini-project logic banayein jisme user se Username, Password, aur ek Security Question poochein nested loops aur if-else ke saath.
